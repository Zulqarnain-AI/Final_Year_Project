```python
"""
Lung Disease Detection using Respiratory Sound Analysis
Dataset: Asthma Detection Dataset Version 2 (Kaggle)

Classes:
0 - Asthma
1 - COPD
2 - Bronchial
3 - Pneumonia
4 - Healthy
"""

import os
import numpy as np
import librosa
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix, classification_report
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

# ---------------------------------------------------------
# CONFIGURATION
# ---------------------------------------------------------

DATASET_PATH = "dataset/Audio_Detection_Dataset"

SAMPLE_RATE = 22050
DURATION = 5
SAMPLES_PER_TRACK = SAMPLE_RATE * DURATION

N_MELS = 128
BATCH_SIZE = 32
EPOCHS = 5

CLASS_NAMES = ["Asthma", "COPD", "Bronchial", "Pneumonia", "Healthy"]

# ---------------------------------------------------------
# AUDIO PREPROCESSING
# ---------------------------------------------------------

def preprocess_audio(file_path):

    signal, sr = librosa.load(file_path, sr=SAMPLE_RATE)

    # Fix audio length
    if len(signal) > SAMPLES_PER_TRACK:
        signal = signal[:SAMPLES_PER_TRACK]
    else:
        signal = np.pad(signal, (0, SAMPLES_PER_TRACK - len(signal)))

    # Mel spectrogram
    mel_spec = librosa.feature.melspectrogram(
        y=signal,
        sr=sr,
        n_mels=N_MELS
    )

    mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)

    # Normalize
    mel_spec_db = (mel_spec_db - mel_spec_db.min()) / (mel_spec_db.max() - mel_spec_db.min())

    mel_spec_db = mel_spec_db[..., np.newaxis]

    return mel_spec_db


# ---------------------------------------------------------
# DATASET LOADER
# ---------------------------------------------------------

def load_dataset(dataset_path):

    features = []
    labels = []

    for label, class_name in enumerate(CLASS_NAMES):

        class_folder = os.path.join(dataset_path, class_name)

        for file in os.listdir(class_folder):

            if file.endswith(".wav"):

                path = os.path.join(class_folder, file)

                try:
                    spec = preprocess_audio(path)

                    features.append(spec)
                    labels.append(label)

                except:
                    print("Error:", path)

    return np.array(features), np.array(labels)


# ---------------------------------------------------------
# LOAD DATA
# ---------------------------------------------------------

print("Loading dataset...")

X, y = load_dataset(DATASET_PATH)

print("Dataset shape:", X.shape)

# Train validation split
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# ---------------------------------------------------------
# CLASS WEIGHTS (Fix class imbalance)
# ---------------------------------------------------------

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)

class_weights = dict(enumerate(class_weights))

print("Class Weights:", class_weights)

# ---------------------------------------------------------
# DATASET PIPELINE
# ---------------------------------------------------------

train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train))
train_dataset = train_dataset.shuffle(1000).batch(BATCH_SIZE)

val_dataset = tf.data.Dataset.from_tensor_slices((X_val, y_val))
val_dataset = val_dataset.batch(BATCH_SIZE)


# ---------------------------------------------------------
# MODEL
# ---------------------------------------------------------

def build_model():

    input_layer = tf.keras.layers.Input(shape=(128, 216, 1))

    # Convert grayscale spectrogram → 3 channels
    x = tf.keras.layers.Concatenate()([input_layer, input_layer, input_layer])

    # Resize for ResNet
    x = tf.keras.layers.Resizing(224, 224)(x)

    # Data augmentation
    x = tf.keras.layers.RandomFlip("horizontal")(x)
    x = tf.keras.layers.RandomRotation(0.05)(x)

    base_model = tf.keras.applications.ResNet50(
        weights="imagenet",
        include_top=False,
        input_shape=(224,224,3)
    )

    # Freeze most layers
    for layer in base_model.layers[:-30]:
        layer.trainable = False

    x = base_model(x)

    x = tf.keras.layers.GlobalAveragePooling2D()(x)

    x = tf.keras.layers.Dense(256, activation="relu")(x)
    x = tf.keras.layers.Dropout(0.5)(x)

    output = tf.keras.layers.Dense(5, activation="softmax")(x)

    model = tf.keras.Model(inputs=input_layer, outputs=output)

    return model


model = build_model()

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# ---------------------------------------------------------
# CALLBACKS
# ---------------------------------------------------------

callbacks = [

    tf.keras.callbacks.EarlyStopping(
        patience=5,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ModelCheckpoint(
        "lung_model.h5",
        save_best_only=True
    )
]

# ---------------------------------------------------------
# TRAINING
# ---------------------------------------------------------

history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks
)

# ---------------------------------------------------------
# EVALUATION
# ---------------------------------------------------------

print("\nEvaluating model...")

y_pred_probs = model.predict(X_val)
y_pred = np.argmax(y_pred_probs, axis=1)

cm = confusion_matrix(y_val, y_pred)

print("\nConfusion Matrix\n")
print(cm)

print("\nClassification Report\n")

print(classification_report(
    y_val,
    y_pred,
    target_names=CLASS_NAMES
))


# ---------------------------------------------------------
# CONFUSION MATRIX PLOT
# ---------------------------------------------------------

plt.figure(figsize=(8,6))
plt.imshow(cm)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.xticks(range(len(CLASS_NAMES)), CLASS_NAMES, rotation=45)
plt.yticks(range(len(CLASS_NAMES)), CLASS_NAMES)
plt.colorbar()
plt.show()


# ---------------------------------------------------------
# INFERENCE
# ---------------------------------------------------------

def predict_disease(audio_path, model):

    spec = preprocess_audio(audio_path)

    spec = np.expand_dims(spec, axis=0)

    prediction = model.predict(spec)

    index = np.argmax(prediction)

    label = CLASS_NAMES[index]

    confidence = np.max(prediction)

    print("Prediction:", label)
    print("Confidence:", confidence)

    return label


# Example
# predict_disease("test.wav", model)
```
